```python
LEGOAI_COMMIT = "<40-hex>"
INTERNET_ON = True  # Confirm the notebook setting before proceeding.
```

```python
import json
import os
import re
import subprocess
import sys
from pathlib import Path

if not INTERNET_ON:
    raise RuntimeError('Internet ON required; no offline dataset path')
if not re.fullmatch(r'[0-9a-f]{40}', LEGOAI_COMMIT):
    raise RuntimeError('PIN_MISSING')
checkout = Path('/tmp/legoai-c1/legoai')
checkout.mkdir(parents=True, exist_ok=False)
fetch_env = {'PATH': '/usr/bin:/bin', 'HOME': '/tmp/legoai-c1/home', 'GIT_TERMINAL_PROMPT': '0', 'GIT_CONFIG_NOSYSTEM': '1', 'GIT_CONFIG_GLOBAL': os.devnull}
commands = [['git', 'init'], ['git', 'remote', 'add', 'origin', 'https://github.com/nishanthrjn/init-Lego-AI-Operating-Layer.git'], ['git', 'fetch', '--depth', '1', 'origin', LEGOAI_COMMIT], ['git', 'checkout', '--detach', 'FETCH_HEAD']]
for argv in commands:
    subprocess.run(argv, cwd=checkout, env=fetch_env, check=True, timeout=180, stdin=subprocess.DEVNULL, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
head = subprocess.run(['git', 'rev-parse', 'HEAD'], cwd=checkout, env=fetch_env, check=True, timeout=15, capture_output=True, text=True).stdout.strip()
if head != LEGOAI_COMMIT:
    raise RuntimeError('SOURCE_IDENTITY_FAILED')
runner = checkout / 'experiments/kaggle/comfyui_t2i_job_1/job_runner.py'
def phase(name):
    subprocess.run([sys.executable, str(runner), name, '--legoai-commit', LEGOAI_COMMIT, '--internet-on'], cwd=checkout, check=True, timeout=1900)
```

```python
phase('inspect')
```

```python
phase('bootstrap')
```

```python
phase('start')
```

```python
phase('generate')
```

```python
phase('finish')
```
