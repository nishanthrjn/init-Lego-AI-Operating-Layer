```python
import json
import subprocess
import sys
from pathlib import Path
sys.dont_write_bytecode = True
TOOLS = Path('/kaggle/input/datasets/REPLACE/local-eval/local_eval')
AUTHORIZATION = ''
CPU_ONLY_CONFIRMED = False
INTERNET_OFF_CONFIRMED = False
OFFICIAL_ENVIRONMENT_AUTHORIZED = False
OUTPUT = '/kaggle/working/localization-study-attempt1'
if AUTHORIZATION != 'AUTHORIZE FORGEAI GEMMA4 LOCALIZATION STUDY CPU' or not all((CPU_ONLY_CONFIRMED, INTERNET_OFF_CONFIRMED, OFFICIAL_ENVIRONMENT_AUTHORIZED)):
    raise RuntimeError('CPU study approval required')
if Path('/dev/nvidia0').exists():
    raise RuntimeError('Accelerator must be None')
sys.path.insert(0, str(TOOLS))
from runtime import clean_env
config_path = Path('/kaggle/working/localization-config.json')
config = dict(authorization=AUTHORIZATION, cpu_only_confirmed=CPU_ONLY_CONFIRMED, internet_off_confirmed=INTERNET_OFF_CONFIRMED, official_environment_authorized=OFFICIAL_ENVIRONMENT_AUTHORIZED, data='/kaggle/input/competitions/gemma-4-developer-agent', output=OUTPUT)
with config_path.open('x') as stream:
    json.dump(config, stream)
subprocess.run([sys.executable, '-B', str(TOOLS / 'localization.py'), '--config', str(config_path)], env=clean_env(), check=True, timeout=11400)
print('Study outputs:', OUTPUT)
```
